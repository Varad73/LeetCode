```python
# 1. Import Libraries
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans


# 2. Load Dataset
df = pd.read_csv("5-sales_data_sample.csv", encoding="unicode_escape")

print("First 5 Rows:")
print(df.head())

print("\nDataset Shape:", df.shape)


# 3. Visualize Deal Size Distribution
df['DEALSIZE'].value_counts().plot(kind='bar')

plt.title("Deal Size Distribution")
plt.xlabel("Deal Size")
plt.ylabel("Count")
plt.show()


# 4. Remove Missing and Duplicate Values
df = df.dropna(subset=['SALES', 'QUANTITYORDERED', 'PRICEEACH'])
df = df.drop_duplicates()


# 5. Select Features
X = df[['SALES', 'QUANTITYORDERED', 'PRICEEACH']]


# 6. Feature Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


# 7. Elbow Method
inertia = []
K = range(1, 12)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)


# 8. Plot Elbow Graph
plt.figure(figsize=(8, 5))

plt.plot(K, inertia, marker='o')

plt.title("Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.xticks(K)
plt.grid(True)

plt.show()


# 9. Apply K-Means Clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

df['Cluster'] = kmeans.fit_predict(X_scaled)


# 10. Visualize Clusters
plt.figure(figsize=(8, 5))

plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=df['Cluster']
)

plt.title("K-Means Clustering")
plt.xlabel("Sales (Scaled)")
plt.ylabel("Quantity Ordered (Scaled)")
plt.show()


# 11. Display Cluster Distribution
print("\nCluster Distribution:")
print(df['Cluster'].value_counts().sort_index())


# 12. Display Cluster Centers
print("\nCluster Centers:")
print(kmeans.cluster_centers_)


# 13. Display Final Dataset
print("\nFinal Clustered Dataset:")
print(df.head())

print("\nTotal Clusters:", df['Cluster'].nunique())
print("Total Records:", len(df))
```
